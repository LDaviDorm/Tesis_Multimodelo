```python
import pandas as pd
from pathlib import Path

# Rutas
entrada = Path("../04_Limpieza/06_Finales/Canonico_FINAL.csv")
salida = Path("../04_Limpieza/06_Finales/Canonico_FINAL.xlsx")

# Leer CSV
df = pd.read_csv(entrada)

# Convertir a números enteros
df["indice"] = pd.to_numeric(df["indice"]).astype(int)
df["Año"] = pd.to_numeric(df["Año"]).astype(int)

# Guardar como Excel
df.to_excel(salida, index=False)

print(f"Archivo creado: {salida}")
print(df[["indice", "Año"]].dtypes)
```
