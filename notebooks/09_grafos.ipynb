```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.lines as mlines

# 1. Crear el grafo
G = nx.Graph()

# 2. Definir nodos
nodos_autor = ["Autor A", "Autor B", "Autor C"]
nodos_afil = ["Afiliación X", "Afiliación Y"]
art_alpha = r"Artículo $\alpha$"
art_beta = r"Artículo $\beta$"
nodos_art = [art_alpha, art_beta]
nodos_kw = [f"KW {i}" for i in range(1, 11)]

for n in nodos_autor: G.add_node(n, tipo="Autor")
for n in nodos_afil: G.add_node(n, tipo="Afiliación")
for n in nodos_art: G.add_node(n, tipo="Artículo")
for n in nodos_kw: G.add_node(n, tipo="Keyword")

# 3. Definir aristas y sus relaciones
edges_autor_art = [("Autor A", art_alpha), ("Autor B", art_alpha), ("Autor A", art_beta), ("Autor C", art_beta)]
edges_autor_afil = [("Autor A", "Afiliación X"), ("Autor B", "Afiliación X"), ("Autor C", "Afiliación Y")]
edges_art_kw = [
    (art_alpha, "KW 1"), (art_alpha, "KW 2"), (art_alpha, "KW 3"), (art_alpha, "KW 4"), (art_alpha, "KW 5"),
    (art_beta, "KW 4"), (art_beta, "KW 5"), (art_beta, "KW 6"), (art_beta, "KW 7"), (art_beta, "KW 8"), 
    (art_beta, "KW 9"), (art_beta, "KW 10")
]

for u, v in edges_autor_art: G.add_edge(u, v, relacion="Autor - Artículo (Escribe)")
for u, v in edges_autor_afil: G.add_edge(u, v, relacion="Autor - Afiliación (Pertenece)")
for u, v in edges_art_kw: G.add_edge(u, v, relacion="Artículo - Keyword (Tiene)")

# 4. Paletas de colores
color_map = {
    "Autor": "#5DADE2",      
    "Artículo": "#E74C3C",   
    "Afiliación": "#48C9B0", 
    "Keyword": "#F4D03F"     
}
edge_color_map = {
    "Autor - Artículo (Escribe)": "#2980B9",         
    "Autor - Afiliación (Pertenece)": "#27AE60",     
    "Artículo - Keyword (Tiene)": "#E67E22"          
}

plt.figure(figsize=(14, 10), facecolor='#F4F6F7')
ax = plt.gca()
ax.set_facecolor('#F4F6F7')

pos = nx.spring_layout(G, seed=42, k=1.2, iterations=200)

# 5. Tamaños ajustados para correcta visualización
size_map = {"Autor": 4800, "Artículo": 6500, "Afiliación": 8000, "Keyword": 1800}
node_colors = [color_map[G.nodes[n]['tipo']] for n in G.nodes()]
node_sizes = [size_map[G.nodes[n]['tipo']] for n in G.nodes()]
edge_colors = [edge_color_map[G[u][v]['relacion']] for u, v in G.edges()]

# 6. Dibujar aristas y nodos
nx.draw_networkx_edges(G, pos, edge_color=edge_colors, width=3.5, alpha=0.85)
nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=node_sizes, edgecolors='#2C3E50', linewidths=2.5)

# Etiquetas
labels_kw = {n: n for n in nodos_kw}
labels_main = {n: n for n in G.nodes() if n not in nodos_kw}
nx.draw_networkx_labels(G, pos, labels=labels_main, font_size=12, font_weight="bold", font_color="#2C3E50")
nx.draw_networkx_labels(G, pos, labels=labels_kw, font_size=10, font_weight="bold", font_color="#2C3E50")

# 7. Leyenda ÚNICA (Tipos de Nodos) en la esquina superior derecha
legend_handles_nodes = [
    mlines.Line2D([], [], color='#F4F6F7', marker='o', markerfacecolor=color, 
                  markeredgecolor='#2C3E50', markeredgewidth=1.5, markersize=14, label=tipo)
    for tipo, color in color_map.items()
]

plt.legend(handles=legend_handles_nodes, loc='upper right', title="Tipos de Nodos", 
           fontsize=11, title_fontsize=13, frameon=True, shadow=True, facecolor='white', edgecolor='#BDC3C7')

plt.title("Grafo principal - relaciones", fontsize=22, fontweight='bold', color='#2C3E50', pad=20)
plt.axis('off')

# Guardar imagen generada
jpg_path = "grafo_relaciones_final.jpg"
plt.savefig(jpg_path, format="jpg", bbox_inches="tight", dpi=300, facecolor='#F4F6F7')
plt.close()
```
