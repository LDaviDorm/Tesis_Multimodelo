```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.lines as mlines

# 1. Crear el grafo
G = nx.Graph()

# 2. Definir nodos
nodos_autor = ["Autor A", "Autor B", "Autor C"]
nodos_afil = ["Afiliación X", "Afiliación Y"]
art_alpha = r"Artículo $\alpha$"
art_beta = r"Artículo $\beta$"
nodos_art = [art_alpha, art_beta]
nodos_kw = [f"KW {i}" for i in range(1, 11)]

for n in nodos_autor: G.add_node(n, tipo="Autor")
for n in nodos_afil: G.add_node(n, tipo="Afiliación")
for n in nodos_art: G.add_node(n, tipo="Artículo")
for n in nodos_kw: G.add_node(n, tipo="Keyword")

# 3. Definir aristas y sus relaciones
edges_autor_art = [("Autor A", art_alpha), ("Autor B", art_alpha), ("Autor A", art_beta), ("Autor C", art_beta)]
edges_autor_afil = [("Autor A", "Afiliación X"), ("Autor B", "Afiliación X"), ("Autor C", "Afiliación Y")]
edges_art_kw = [
    (art_alpha, "KW 1"), (art_alpha, "KW 2"), (art_alpha, "KW 3"), (art_alpha, "KW 4"), (art_alpha, "KW 5"),
    (art_beta, "KW 4"), (art_beta, "KW 5"), (art_beta, "KW 6"), (art_beta, "KW 7"), (art_beta, "KW 8"), 
    (art_beta, "KW 9"), (art_beta, "KW 10")
]

for u, v in edges_autor_art: G.add_edge(u, v, relacion="Autor - Artículo (Escribe)")
for u, v in edges_autor_afil: G.add_edge(u, v, relacion="Autor - Afiliación (Pertenece)")
for u, v in edges_art_kw: G.add_edge(u, v, relacion="Artículo - Keyword (Tiene)")

# 4. Paletas de colores
color_map = {
    "Autor": "#5DADE2",      
    "Artículo": "#E74C3C",   
    "Afiliación": "#48C9B0", 
    "Keyword": "#F4D03F"     
}
edge_color_map = {
    "Autor - Artículo (Escribe)": "#2980B9",         
    "Autor - Afiliación (Pertenece)": "#27AE60",     
    "Artículo - Keyword (Tiene)": "#E67E22"          
}

plt.figure(figsize=(14, 10), facecolor='#F4F6F7')
ax = plt.gca()
ax.set_facecolor('#F4F6F7')

pos = nx.spring_layout(G, seed=42, k=1.2, iterations=200)

# 5. Tamaños ajustados para correcta visualización
size_map = {"Autor": 4800, "Artículo": 6500, "Afiliación": 8000, "Keyword": 1800}
node_colors = [color_map[G.nodes[n]['tipo']] for n in G.nodes()]
node_sizes = [size_map[G.nodes[n]['tipo']] for n in G.nodes()]
edge_colors = [edge_color_map[G[u][v]['relacion']] for u, v in G.edges()]

# 6. Dibujar aristas y nodos
nx.draw_networkx_edges(G, pos, edge_color=edge_colors, width=3.5, alpha=0.85)
nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=node_sizes, edgecolors='#2C3E50', linewidths=2.5)

# Etiquetas
labels_kw = {n: n for n in nodos_kw}
labels_main = {n: n for n in G.nodes() if n not in nodos_kw}
nx.draw_networkx_labels(G, pos, labels=labels_main, font_size=12, font_weight="bold", font_color="#2C3E50")
nx.draw_networkx_labels(G, pos, labels=labels_kw, font_size=10, font_weight="bold", font_color="#2C3E50")

# 7. Leyenda ÚNICA (Tipos de Nodos) en la esquina superior derecha
legend_handles_nodes = [
    mlines.Line2D([], [], color='#F4F6F7', marker='o', markerfacecolor=color, 
                  markeredgecolor='#2C3E50', markeredgewidth=1.5, markersize=14, label=tipo)
    for tipo, color in color_map.items()
]

plt.legend(handles=legend_handles_nodes, loc='upper right', title="Tipos de Nodos", 
           fontsize=11, title_fontsize=13, frameon=True, shadow=True, facecolor='white', edgecolor='#BDC3C7')

plt.title("Grafo principal - relaciones", fontsize=22, fontweight='bold', color='#2C3E50', pad=20)
plt.axis('off')

# Guardar imagen generada
jpg_path = "grafo_relaciones_final.jpg"
plt.savefig(jpg_path, format="jpg", bbox_inches="tight", dpi=300, facecolor='#F4F6F7')
plt.close()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.lines as mlines

# 1. Crear el grafo
G = nx.Graph()

# 2. Definir nodos, etiquetas genéricas y atributos (afiliaciones)
nodos_info = {
    "A1": {"label": "Autor 1", "afil": "Afiliación 1: Instituto de Ingeniería\nAfiliación 2: CCADET"},
    "A2": {"label": "Autor 2", "afil": "Afiliación 1: IIMAS\nAfiliación 2: Facultad de Ciencias"},
    "A3": {"label": "Autor 3", "afil": "Afiliación 1: Facultad de Ciencias\nAfiliación 2: --"},
    "A4": {"label": "Autor 4", "afil": "Afiliación 1: FES Acatlán\nAfiliación 2: --"},
    "A5": {"label": "Autor 5", "afil": "Afiliación 1: IIMAS\nAfiliación 2: Instituto de Ingeniería"}
}

G.add_nodes_from(nodos_info.keys())

# 3. Definir aristas (coautorías)
edges = [("A1", "A2"), ("A1", "A3"), ("A2", "A3"), ("A2", "A4"), ("A3", "A5"), ("A4", "A5")]
G.add_edges_from(edges)

# 4. Configurar lienzo
plt.figure(figsize=(14, 10), facecolor='#F4F6F7')
ax = plt.gca()
ax.set_facecolor('#F4F6F7')

# Posiciones fijas
pos = {
    "A2": (0, 1),
    "A1": (-1, 0.2),
    "A4": (1, 0.2),
    "A3": (-0.5, -0.8),
    "A5": (0.5, -0.8)
}

# 5. Dibujar Aristas y Nodos
color_aristas = "#7F8C8D"  # Gris contrastante
nx.draw_networkx_edges(G, pos, edge_color=color_aristas, width=3.5, alpha=0.85)

color_nodos = "#5DADE2"    # Azul
nx.draw_networkx_nodes(G, pos, node_color=color_nodos, node_size=7500, edgecolors='#2C3E50', linewidths=2.5)

# 6. Etiquetas de los nodos
labels = {n: data["label"] for n, data in nodos_info.items()}
nx.draw_networkx_labels(G, pos, labels=labels, font_size=12, font_weight="bold", font_color="#2C3E50")

# 7. Cajas de atributos
for n, (x, y) in pos.items():
    plt.text(x, y - 0.24, nodos_info[n]["afil"], fontsize=9.5, ha='center', va='top', color='#2C3E50',
             bbox=dict(boxstyle="round,pad=0.5", fc="#FFFFFF", ec="#BDC3C7", lw=1.5, alpha=0.9))

# 8. Leyenda
legend_handles = [
    mlines.Line2D([], [], color='#F4F6F7', marker='o', markerfacecolor=color_nodos, 
                  markeredgecolor='#2C3E50', markeredgewidth=1.5, markersize=14, label="Autor (Nodo)"),
    mlines.Line2D([], [], color=color_aristas, linewidth=3.5, label="Coautoría (Arista)")
]

plt.legend(handles=legend_handles, loc='upper right', title="Elementos del Subgrafo", 
           fontsize=11, title_fontsize=13, frameon=True, shadow=True, facecolor='white', edgecolor='#BDC3C7')

plt.title("Subgrafo de coautoría", fontsize=22, fontweight='bold', color='#2C3E50', pad=20)

plt.xlim(-1.5, 1.5)
plt.ylim(-1.3, 1.3)
plt.axis('off')

# Guardar
plt.savefig("subgrafo_coautoria_generico.jpg", format="jpg", bbox_inches="tight", dpi=300, facecolor='#F4F6F7')
plt.close()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.lines as mlines
from matplotlib.patches import FancyArrowPatch

# ====================================================
# 1. CREAR GRAFO DE EJEMPLO PARA LA CONSULTA
# ====================================================

G = nx.Graph()

articulo = "Artículo α"
autores = ["Autor A", "Autor B"]
keywords = ["KW 1", "KW 2", "KW 3"]

G.add_node(articulo, tipo="Artículo")
for a in autores:
    G.add_node(a, tipo="Autor")
for kw in keywords:
    G.add_node(kw, tipo="Keyword")

# Aristas del artículo con autores y keywords
for a in autores:
    G.add_edge(articulo, a, relacion="Autor - Artículo")

for kw in keywords:
    G.add_edge(articulo, kw, relacion="Artículo - Keyword")

# ====================================================
# 2. PALETA DE COLORES Y ESTILO
# ====================================================

bg_color = "#F4F6F7"
border_color = "#2C3E50"

color_map = {
    "Autor": "#5DADE2",
    "Artículo": "#E74C3C",
    "Keyword": "#F4D03F"
}

edge_color_map = {
    "Autor - Artículo": "#2980B9",
    "Artículo - Keyword": "#E67E22"
}

# ====================================================
# 3. FIGURA Y POSICIONES
# ====================================================

fig, ax = plt.subplots(figsize=(16, 8), facecolor=bg_color)
ax.set_facecolor(bg_color)

# Posiciones fijas para mantener un diseño claro
pos = {
    articulo: (-2.2, 0.0),
    "Autor A": (-3.6, 1.1),
    "Autor B": (-3.6, -1.1),
    "KW 1": (-0.8, 1.4),
    "KW 2": (-0.3, 0.0),
    "KW 3": (-0.8, -1.4),
}

# ====================================================
# 4. DIBUJAR SUBGRAFO DE CONSULTA
# ====================================================

# Aristas
edge_colors = [edge_color_map[G[u][v]["relacion"]] for u, v in G.edges()]
nx.draw_networkx_edges(
    G, pos,
    ax=ax,
    edge_color=edge_colors,
    width=3.5,
    alpha=0.9
)

# Nodos
size_map = {
    "Autor": 4200,
    "Artículo": 6500,
    "Keyword": 1800
}

node_colors = [color_map[G.nodes[n]["tipo"]] for n in G.nodes()]
node_sizes = [size_map[G.nodes[n]["tipo"]] for n in G.nodes()]

nx.draw_networkx_nodes(
    G, pos,
    ax=ax,
    node_color=node_colors,
    node_size=node_sizes,
    edgecolors=border_color,
    linewidths=2.5
)

# Etiquetas
labels_main = {n: n for n in G.nodes()}
nx.draw_networkx_labels(
    G, pos,
    labels=labels_main,
    ax=ax,
    font_size=13,
    font_weight="bold",
    font_color=border_color
)

# ====================================================
# 5. RESALTAR EL PASO DE CONSULTA
# ====================================================

# Caja descriptiva izquierda
ax.text(
    -2.2, 2.15,
    "Paso 1: Seleccionar un artículo\n"
    "Paso 2: Obtener sus vecinos\n"
    "Paso 3: Filtrar por tipo = autor",
    ha="center",
    va="center",
    fontsize=12,
    color=border_color,
    bbox=dict(
        boxstyle="round,pad=0.5",
        fc="white",
        ec="#BDC3C7",
        lw=1.5,
        alpha=0.96
    )
)

# Flecha horizontal al resultado
arrow = FancyArrowPatch(
    posA=(-0.05, 0.0),
    posB=(1.8, 0.0),
    arrowstyle='simple',
    mutation_scale=25,
    lw=0,
    color="#27AE60",
    alpha=0.9
)
ax.add_patch(arrow)

# Texto sobre flecha
ax.text(
    0.9, 0.42,
    "Filtrar vecinos\ncon tipo = autor",
    ha="center",
    va="center",
    fontsize=12,
    fontweight="bold",
    color="#1E8449"
)

# ====================================================
# 6. RESULTADO DE LA CONSULTA
# ====================================================

# Caja de resultado
resultado_texto = (
    "Resultado de la consulta\n\n"
    "Autores encontrados:\n"
    "• Autor A\n"
    "• Autor B"
)

ax.text(
    3.3, 0.0,
    resultado_texto,
    ha="center",
    va="center",
    fontsize=14,
    color=border_color,
    bbox=dict(
        boxstyle="round,pad=0.8",
        fc="white",
        ec="#BDC3C7",
        lw=1.8,
        alpha=0.98
    )
)

# Etiqueta pequeña abajo del resultado
ax.text(
    3.3, -1.55,
    "Se conservan únicamente los vecinos\n"
    "correspondientes a autores.",
    ha="center",
    va="center",
    fontsize=11,
    color=border_color
)

# ====================================================
# 7. LEYENDA
# ====================================================

legend_handles = [
    mlines.Line2D(
        [], [], color=bg_color, marker='o',
        markerfacecolor=color_map["Autor"],
        markeredgecolor=border_color,
        markeredgewidth=1.5,
        markersize=14,
        label="Autor"
    ),
    mlines.Line2D(
        [], [], color=bg_color, marker='o',
        markerfacecolor=color_map["Artículo"],
        markeredgecolor=border_color,
        markeredgewidth=1.5,
        markersize=14,
        label="Artículo"
    ),
    mlines.Line2D(
        [], [], color=bg_color, marker='o',
        markerfacecolor=color_map["Keyword"],
        markeredgecolor=border_color,
        markeredgewidth=1.5,
        markersize=14,
        label="Keyword"
    ),
    mlines.Line2D(
        [], [], color="#27AE60",
        linewidth=6,
        label="Paso de filtrado"
    )
]

plt.legend(
    handles=legend_handles,
    loc='upper right',
    title="Elementos de la consulta",
    fontsize=11,
    title_fontsize=13,
    frameon=True,
    shadow=True,
    facecolor='white',
    edgecolor='#BDC3C7'
)

# ====================================================
# 8. TÍTULO Y AJUSTES FINALES
# ====================================================

plt.title(
    "Consulta en el grafo principal",
    fontsize=22,
    fontweight='bold',
    color=border_color,
    pad=18
)

ax.set_xlim(-5.0, 5.2)
ax.set_ylim(-2.3, 2.7)
ax.axis("off")

# Guardar
out_path = "consulta_autores_grafo_principal.png"
plt.savefig(
    out_path,
    format="png",
    bbox_inches="tight",
    dpi=300,
    facecolor=bg_color
)
plt.close()

print(f"Imagen guardada en: {out_path}")
```
